```python
import pandas as pd

```

```python
df = pd.read_csv('../data/application_train.csv')

df.head()
```

```python
cols = [
    'TARGET',
    'AMT_INCOME_TOTAL',
    'AMT_CREDIT',
    'DAYS_EMPLOYED',
    'DAYS_BIRTH',
    'CNT_CHILDREN',
    'CODE_GENDER',
    'NAME_EDUCATION_TYPE',
    'NAME_FAMILY_STATUS',
    'FLAG_OWN_CAR',
    'FLAG_OWN_REALTY',
    'AMT_ANNUITY',        
    'REGION_POPULATION_RELATIVE', 
    'DAYS_REGISTRATION'
]
```

```python
df = df[cols]
df.head()
```

```python
print(df.shape)
```

```python
df['DEBT_TO_INCOME'] = df['AMT_CREDIT']/df['AMT_INCOME_TOTAL']
```

```python
df['AGE'] = ((-df['DAYS_BIRTH'])/365).round(1)
```

```python
df['EMPLOYMENT_YEARS'] = ((-df['DAYS_EMPLOYED'])/365).round(1)
```

```python
df['EMPLOYMENT_YEARS'] = df['EMPLOYMENT_YEARS'].replace(1000.7, None)
df['DAYS_EMPLOYED'] = df['DAYS_EMPLOYED'].replace(365243, None)
```

```python
df = df.drop(['DAYS_BIRTH', 'DAYS_EMPLOYED', 'AMT_CREDIT'], axis=1)
```

```python

df['EMPLOYMENT_YEARS'] = df['EMPLOYMENT_YEARS'].fillna(df['EMPLOYMENT_YEARS'].median())
df['DEBT_TO_INCOME'] = df['DEBT_TO_INCOME'].fillna(df['DEBT_TO_INCOME'].median())


print(df.isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
text_cols = ['CODE_GENDER', 'NAME_EDUCATION_TYPE', 'NAME_FAMILY_STATUS', 
             'FLAG_OWN_CAR', 'FLAG_OWN_REALTY']
```

```python
for cols in text_cols:
    df[cols] = le.fit_transform(df[cols].astype(str))
```

```python
df.head()
```

```python
df.to_csv('../data/clean_data.csv', index=False)
```
